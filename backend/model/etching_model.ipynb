```python
df_train = pd.read_csv("../data/modeling/train_encoded.csv")
df_test = pd.read_csv("../data/modeling/test_encoded.csv")

display(df_train.head())
print("train shape:", df_train.shape)
print("test shape:", df_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

etching = ['Thin F4', 'Thin F3', 'Thin F2', 'Thin F1', 'Temp_Etching', 'Selectivity', 'Source_Power']
y = 'is_low_yield'

X_train = df_train[etching].copy()
X_test  = df_test[etching].copy()
y_train = df_train[y].copy()
y_test  = df_test[y].copy()

model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    class_weight=None,
    max_depth=None,
    max_features='sqrt',
    min_samples_leaf=1,
    min_samples_split=2,
    n_estimators=400,
)
model.fit(X_train, y_train)

print(classification_report(y_test, model.predict(X_test), zero_division=0))
```

```python
import joblib
from pathlib import Path

save_path = Path("../model/etching_rf.joblib")
save_path.parent.mkdir(parents=True, exist_ok=True)
joblib.dump(model, save_path)
print(f"모델 저장 완료: {save_path.resolve()}")
```
